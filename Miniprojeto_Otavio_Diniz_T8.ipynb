## CP1.2 — Diagnóstico da qualidade dos dados

### O que vamos investigar?

No checkpoint anterior, conhecemos a estrutura da Base Varejo. Agora precisamos verificar a qualidade das informações recebidas.

Uma base pode conter problemas que não aparecem quando observamos apenas as primeiras cinco linhas. Por isso, vamos investigar todo o conjunto de dados antes de realizar qualquer alteração.

**Pergunta principal:** existem informações ausentes, registros repetidos ou diferenças na escrita dos textos que possam comprometer nossa análise?

### O que será verificado?

**1. Valores ausentes**

Usaremos `isna()` para identificar campos sem informação. Em Pandas, esses valores podem aparecer como `NaN`.

**2. Textos vazios e espaços**

Um campo pode conter somente espaços ou ter espaços desnecessários antes ou depois do texto. Vamos verificar essas situações utilizando `str.strip()`.

**3. Duplicatas**

Usaremos `duplicated()` para identificar linhas completamente repetidas.

É importante lembrar que um cliente pode realizar várias compras e que uma compra pode conter vários produtos. Por isso, uma repetição não deve ser removida sem investigar seu significado.

**4. Inconsistências textuais**

Vamos observar as categorias e os textos existentes para identificar diferenças de escrita ou marcadores como `#N/D`.

### Conhecimentos utilizados

Neste checkpoint, aplicaremos recursos estudados anteriormente:

- Variáveis, listas e laços `for` do Python.
- Condicionais e leitura de resultados.
- DataFrames e seleção de colunas.
- Métodos `isna()`, `sum()`, `duplicated()`, `unique()` e operações com textos do Pandas.

**Dica de Ouro:** identificar um problema não significa que devemos corrigi-lo imediatamente. Primeiro precisamos conhecer sua quantidade, sua natureza e seu possível impacto. As decisões de limpeza serão realizadas na Sprint 2.

````python
# CP1.2 — Verificação de valores ausentes

print("Valores ausentes por coluna:")

ausentes = df_bruto.isna().sum()

display(ausentes)
````

````python
# CP1.2 — Verificação dos campos de texto

colunas_texto = [
    "DATA",
    "CL_GENERO",
    "CL_SEG",
    "PR_CAT",
    "PR_NOME"
]

for coluna in colunas_texto:

    texto = df_bruto[coluna].astype("string")

    # Campos vazios ou contendo apenas espaços
    em_branco = texto.str.strip().eq("").sum()

    # Valores com espaços no início ou no fim
    com_espacos = texto.ne(texto.str.strip()).sum()

    print(
        coluna,
        "- Em branco:", em_branco,
        "- Com espaços nas extremidades:", com_espacos
    )
````

````python
# CP1.2 — Verificação de registros repetidos

print("Linhas completamente duplicadas:")
print(df_bruto.duplicated().sum())

print("\nIdentificadores distintos:")

print(
    "Compras:",
    df_bruto["CO_ID"].nunique()
)

print(
    "Clientes:",
    df_bruto["CL_ID"].nunique()
)

print(
    "Produtos:",
    df_bruto["PR_ID"].nunique()
)
````

````python
# CP1.2 — Conferência das categorias textuais

for coluna in ["CL_GENERO", "CL_SEG", "PR_CAT"]:

    print("\nColuna:", coluna)

    print(df_bruto[coluna].unique())


# Conferir marcadores textuais de informação ausente
print("\nOcorrências do marcador #N/D:")

for coluna in ["PR_CAT", "PR_NOME"]:

    quantidade = (
        df_bruto[coluna]
        .astype("string")
        .str.strip()
        .eq("#N/D")
        .sum()
    )

    print(coluna, ":", quantidade)


# Observar alguns nomes de produtos
print("\nExemplos de nomes de produtos:")

display(
    df_bruto["PR_NOME"]
    .value_counts()
    .head(10)
)
````

# Sprint 2 — Limpeza e padronização dos dados

## CP2.1 — Investigação e decisões de tratamento

### Objetivo

Na Sprint 1, identificamos possíveis problemas de qualidade na Base Varejo. Agora precisamos avaliar quais situações realmente exigem correção e quais informações devem ser preservadas.

O objetivo deste checkpoint é **investigar e justificar as decisões antes de modificar os dados**.

### O que foi identificado anteriormente?

- Quatro colunas sem valores preenchidos.
- 96.553 repetições completas de linhas.
- 3.650 ocorrências do marcador `#N/D` em cada uma das colunas `PR_CAT` e `PR_NOME`.
- Necessidade de compreender melhor as relações entre compras, clientes e produtos.

### O que vamos investigar?

**1. Repetições de registros**

Vamos observar exemplos de linhas repetidas e verificar se existem elementos suficientes para considerar essas ocorrências erros.

**2. Relações entre os identificadores**

Vamos investigar se uma compra está associada a diferentes clientes ou datas e se os identificadores de clientes e produtos apresentam informações divergentes.

**3. Marcadores de informação indisponível**

Vamos verificar se as ocorrências de `#N/D` em `PR_CAT` e `PR_NOME` aparecem nos mesmos registros.

**4. Decisões de limpeza**

Ao final, registraremos quais problemas poderão ser corrigidos, quais informações precisam ser preservadas e quais limitações permanecerão documentadas.

### Conceitos utilizados

Aplicaremos métodos já estudados em Pandas, como `duplicated()`, `groupby()`, `nunique()`, `sum()` e `loc[]`.

**Dica de Ouro:** uma boa limpeza de dados não consiste em eliminar o maior número possível de registros. Consiste em melhorar a confiabilidade das informações sem perder dados válidos.

````python
# CP2.1 — Investigação das linhas repetidas

# Identificar todas as ocorrências de linhas repetidas
repetidas = df_bruto.duplicated(keep=False)

print("Linhas pertencentes a grupos repetidos:")
print(repetidas.sum())

print("\nRepetições excedentes:")
print(df_bruto.duplicated().sum())

# Observar alguns exemplos
colunas_analise = [
    "DATA", "CO_ID", "CL_ID",
    "PR_ID", "PR_CAT", "PR_NOME"
]

display(
    df_bruto.loc[repetidas, colunas_analise].head(10)
)
````

````python
# CP2.1 — Consistência das informações

verificacoes = [
    ("Compras com mais de um cliente",
     "CO_ID", "CL_ID"),

    ("Compras com mais de uma data",
     "CO_ID", "DATA"),

    ("Clientes com mais de um gênero registrado",
     "CL_ID", "CL_GENERO"),

    ("Clientes com mais de um segmento",
     "CL_ID", "CL_SEG"),

    ("Produtos com mais de um nome",
     "PR_ID", "PR_NOME"),

    ("Produtos com mais de uma categoria",
     "PR_ID", "PR_CAT")
]

for descricao, identificador, campo in verificacoes:

    quantidades = (
        df_bruto.groupby(identificador)[campo].nunique()
    )

    divergencias = (quantidades > 1).sum()

    print(descricao, ":", divergencias)
````

````python
# CP2.1 — Investigação dos marcadores #N/D

categoria_sem_dado = (
    df_bruto["PR_CAT"].astype("string").str.strip().eq("#N/D")
)

nome_sem_dado = (
    df_bruto["PR_NOME"].astype("string").str.strip().eq("#N/D")
)

print("Marcadores em PR_CAT:", categoria_sem_dado.sum())
print("Marcadores em PR_NOME:", nome_sem_dado.sum())

print(
    "Marcadores nas duas colunas da mesma linha:",
    (categoria_sem_dado & nome_sem_dado).sum()
)

print(
    "Registros com marcador em pelo menos uma coluna:",
    (categoria_sem_dado | nome_sem_dado).sum()
)

display(
    df_bruto.loc[
        categoria_sem_dado | nome_sem_dado,
        ["CO_ID", "PR_ID", "PR_CAT", "PR_NOME"]
    ].head(10)
)
````

````python
# CP2.1 — Verificações complementares

# Identificar os códigos de produto com marcador #N/D
codigos_sem_dado = df_bruto.loc[
    categoria_sem_dado | nome_sem_dado,
    "PR_ID"
]

print("Códigos de produto associados a #N/D:")
display(codigos_sem_dado.value_counts())

# Verificar se esses códigos possuem outros nomes ou categorias
print("\nInformações disponíveis para esses códigos:")

display(
    df_bruto.loc[
        df_bruto["PR_ID"].isin(codigos_sem_dado.unique()),
        ["PR_ID", "PR_CAT", "PR_NOME"]
    ].drop_duplicates()
)

# Completar a conferência dos atributos do cliente
print("\nConsistência de outros campos do cliente:")

for coluna in ["CL_EC", "CL_FHL"]:

    divergencias = (
        df_bruto.groupby("CL_ID")[coluna]
        .nunique()
        .gt(1)
        .sum()
    )

    print(coluna, "- Clientes com valores diferentes:", divergencias)
````

O método `duplicated(keep=False)` permitiu identificar todas as linhas pertencentes a grupos repetidos, incluindo suas primeiras ocorrências. Já `duplicated()` contabilizou somente as repetições adicionais.

**Interpretação:** apesar da repetição dos registros, ainda não podemos afirmar que todas essas ocorrências sejam erros.

Uma compra pode conter mais de uma unidade do mesmo produto. Como a base não apresenta uma coluna de quantidade nem um identificador individual de item, não é possível distinguir com segurança uma repetição indevida de uma ocorrência legítima.

Por esse motivo, as linhas repetidas serão preservadas até que exista uma justificativa suficiente para sua eventual exclusão.

# Sprint 3 — Funções e organização

## CP3.1 — Compreendendo a função de tratamento

### Objetivo

Nas sprints anteriores, aprendemos a importar, investigar e limpar uma base de dados com Pandas.

Agora vamos organizar parte desse trabalho em funções, para que os mesmos procedimentos possam ser utilizados novamente sem precisar reescrever todos os comandos.

### O que é uma função?

Uma função é um conjunto de instruções criado para realizar uma tarefa específica.

Em Python, utilizamos:

- `def` para definir a função;
- **parâmetro** para receber a informação que será utilizada;
- `return` para devolver o resultado do processamento.

Uma função não precisa modificar diretamente a base original. Podemos receber um DataFrame, criar uma cópia, realizar os tratamentos e devolver a versão tratada.

### Aplicação no mini-projeto

Na Sprint 2, realizamos a limpeza com comandos executados diretamente no notebook.

Na Sprint 3, o objetivo é compreender como reunir esses tratamentos em uma função reutilizável.

**Dica de Ouro:** `print()` mostra uma informação na tela; `return` devolve um resultado que poderá ser armazenado em uma variável e utilizado nas próximas etapas.

````python
# CP3.1 — Organizar o tratamento em uma função

# Criar uma função que recebe uma tabela
def tratar_dados(tabela):

    # Preservar a tabela recebida
    tratado = tabela.copy()

    # PASSO 1 — Remover as quatro colunas vazias
    tratado = tratado.drop(
        columns=[
            "Unnamed: 10",
            "Unnamed: 11",
            "Unnamed: 12",
            "Unnamed: 13"
        ]
    )

    # PASSO 2 — Padronizar a categoria desconhecida
    tratado["PR_CAT"] = tratado["PR_CAT"].replace(
        "#N/D", "NÃO INFORMADA"
    )

    # PASSO 3 — Padronizar o nome desconhecido
    tratado["PR_NOME"] = tratado["PR_NOME"].replace(
        "#N/D", "NÃO INFORMADO"
    )

    # Devolver a tabela tratada
    return tratado


# CONFERÊNCIA — Chamar a função
df_funcao = tratar_dados(df_bruto)

print("Dimensões:", df_funcao.shape)

print(
    "Mesmo resultado do CP2.2:",
    df_funcao.equals(df_trabalho)
)

print("\nPrimeiras linhas da tabela tratada:")
display(df_funcao.head())
````

## CP3.2 — Função de resumo da Base Varejo

### Objetivo

Organizar em uma função reutilizável o resumo da base tratada, preservando a distinção entre registros e identificadores distintos.

### Procedimento

A função `gerar_resumo(tabela)` recebe o DataFrame e retorna um dicionário com quatro indicadores: quantidade de registros, compras distintas (`CO_ID`), clientes distintos (`CL_ID`) e produtos distintos (`PR_ID`). Em seguida, a função é chamada com `df_funcao`, conforme o código apresentado abaixo.


````python
# CP3.2 — Função de resumo da Base Varejo

def gerar_resumo(tabela):

    resumo = {
        "registros": len(tabela),
        "compras_distintas": tabela["CO_ID"].nunique(),
        "clientes_distintos": tabela["CL_ID"].nunique(),
        "produtos_distintos": tabela["PR_ID"].nunique()
    }

    return resumo


# Chamar a função com a base tratada
resumo_varejo = gerar_resumo(df_funcao)

print(resumo_varejo)
````

## CP4.2 — Agrupamentos com `groupby()`

### Objetivo

Agrupar os registros da Base Varejo para comparar categorias de produtos e segmentos de clientes.

### Investigação 1 — Registros por categoria

Identificar quantas linhas da base pertencem a cada categoria de produto, utilizando `PR_CAT`.

### Investigação 2 — Compras distintas por segmento

Identificar quantos códigos de compra diferentes existem em cada segmento, utilizando `CL_SEG` e `CO_ID`.

### Cuidados na interpretação

- `.size()` conta as linhas de cada grupo.
- `.nunique()` conta os valores diferentes de uma coluna dentro de cada grupo.
- A quantidade de registros não deve ser confundida com compras distintas, unidades vendidas ou faturamento.

Os resultados serão interpretados após a execução dos agrupamentos.

````python
# CP4.2 — Agrupamentos da Base Varejo

# 1. Contar registros por categoria de produto
print("Registros por categoria:")

registros_por_categoria = (
    df_funcao.groupby("PR_CAT")
    .size()
    .sort_values(ascending=False)
)

display(registros_por_categoria)


# 2. Contar compras distintas por segmento
print("Compras distintas por segmento:")

compras_por_segmento = (
    df_funcao.groupby("CL_SEG")["CO_ID"]
    .nunique()
    .sort_values(ascending=False)
)

display(compras_por_segmento)
````

**Diferença entre as métricas:** O método `.size()` conta todas as linhas existentes em cada grupo, mesmo quando um código de compra aparece repetidamente. Já o método `.nunique()` conta somente os valores distintos de uma coluna. Neste caso, ele identifica quantos códigos de compra (`CO_ID`) diferentes existem em cada segmento.

Por isso, uma mesma compra pode contribuir com várias linhas para a contagem de registros, mas será contabilizada apenas uma vez na contagem de compras distintas do seu segmento.

### Procedimento

A análise utiliza `groupby()` e `.nunique()` para construir a medida por compra. Em seguida, são aplicados os métodos `.count()`, `.mean()`, `.min()` e `.max()` sobre a série numérica obtida.

````python
# CP4.3 — Estatísticas básicas

tipos_por_compra = (
    df_funcao.groupby("CO_ID")["PR_ID"]
    .nunique()
)

contagem = tipos_por_compra.count()
media = tipos_por_compra.mean()
minimo = tipos_por_compra.min()
maximo = tipos_por_compra.max()

print("Contagem válida:", contagem, "compras")
print("Média:", round(media, 2), "tipos de produtos por compra")
print("Mínimo:", minimo, "tipo(s) de produto por compra")
print("Máximo:", maximo, "tipo(s) de produto por compra")
````

- **Cliente distinto (`CL_ID`):** identifica cada cliente uma única vez dentro do seu grupo de gênero.
- **Compra distinta (`CO_ID`):** identifica cada compra uma única vez dentro do grupo de gênero associado ao cliente.
- **Registro:** representa uma linha da base de dados, que pode conter uma ocorrência de produto associada a determinada compra.

### Objetivo do procedimento

Comparar clientes distintos, compras distintas e registros ajuda a compreender a composição da base em níveis diferentes.

A contagem de clientes informa quantas pessoas distintas existem em cada grupo. A contagem de compras mostra quantas operações de compra distintas estão associadas a esses clientes. Já a contagem de registros revela a quantidade de ocorrências documentadas na base.

Essas medidas não são equivalentes. Um cliente pode estar associado a várias compras, e uma compra pode aparecer em diversas linhas, inclusive com repetições de códigos de produtos.

Portanto, uma quantidade maior de registros não comprova, por si só, maior quantidade física de produtos vendidos ou maior valor gasto.

A comparação será realizada por meio de agrupamentos, contagens de valores distintos e conferência dos totais.


````python
# CP4.4 — Perfil dos clientes por gênero

resumo_genero = (
    df_funcao
    .groupby("CL_GENERO")
    .agg(
        clientes_distintos=("CL_ID", "nunique"),
        compras_distintas=("CO_ID", "nunique"),
        registros=("CL_ID", "size")
    )
    .reset_index()
)

display(resumo_genero)

print("Clientes distintos:", resumo_genero["clientes_distintos"].sum())
print("Compras distintas:", resumo_genero["compras_distintas"].sum())
print("Registros:", resumo_genero["registros"].sum())
````

Portanto, o grupo F apresenta participação um pouco superior ao grupo M em clientes, compras e registros. A diferença é pequena e permanece próxima de quatro pontos percentuais em todos os indicadores, indicando que a composição da base está relativamente equilibrada entre os dois grupos.

**Interpretação das métricas**

A contagem de clientes distintos identifica quantas pessoas diferentes estão representadas em cada grupo.

A contagem de compras distintas identifica quantas operações de compra estão associadas aos clientes desses grupos.

A contagem de registros considera todas as ocorrências presentes na base, inclusive quando uma compra aparece em várias linhas.

Esses indicadores representam dimensões diferentes e não devem ser interpretados como equivalentes.

**Limitações da análise**

Os resultados permitem identificar como clientes, compras e registros estão distribuídos, mas não explicam, isoladamente, as causas das diferenças observadas.

## CP4.5 — Distribuição das categorias de produtos por gênero

### Pergunta de investigação

Como os registros das diferentes categorias de produtos estão distribuídos entre os grupos de gênero M e F?

### Objetivo

Na investigação anterior, foram examinados clientes, compras distintas e registros por gênero.

Agora, o objetivo é aprofundar essa análise por meio do cruzamento entre `PR_CAT` e `CL_GENERO`.

Esse cruzamento permite observar como as categorias de produtos se distribuem dentro de cada grupo, em vez de analisar gênero e categoria separadamente.

### Unidade de análise

A unidade analisada será o **registro da Base Varejo**, contado em cada combinação de categoria de produto e código de gênero.

A contagem corresponde às ocorrências registradas, e não necessariamente à quantidade física de unidades vendidas.

### Procedimento

Serão considerados:

- O número absoluto de registros de cada categoria em cada grupo de gênero.
- O total de registros de cada grupo.
- A participação percentual de cada categoria dentro de seu respectivo grupo.
- A conferência dos totais utilizados nos cálculos.

A comparação percentual deve utilizar o total de registros do próprio grupo de gênero como denominador.

### Limitações

A base conserva registros repetidos cuja exclusão não foi justificada.

Por isso, os resultados devem ser interpretados como distribuição de registros, sem pressupor que cada linha represente uma unidade física vendida.

Além disso, diferenças nas frequências das categorias não comprovam, isoladamente, preferências de consumo ou relações causais entre gênero e comportamento de compra.

````python
# CP4.5 — Distribuição cruzada entre gênero e categoria de produto

distribuicao_genero_categoria = (
    df_funcao
    .groupby(["CL_GENERO", "PR_CAT"])
    .size()
    .reset_index(name="registros")
)

# Total de registros dentro de cada grupo de gênero
total_por_genero = (
    df_funcao
    .groupby("CL_GENERO")
    .size()
    .rename("total_genero")
)

# Acrescentar o total do gênero à tabela
distribuicao_genero_categoria = (
    distribuicao_genero_categoria
    .merge(
        total_por_genero,
        on="CL_GENERO"
    )
)

# Calcular percentuais mantendo a precisão original
distribuicao_genero_categoria["percentual_no_genero"] = (
    distribuicao_genero_categoria["registros"]
    / distribuicao_genero_categoria["total_genero"]
    * 100
)

# Arredondar somente para exibir
tabela_exibicao = distribuicao_genero_categoria.copy()

tabela_exibicao["percentual_no_genero"] = (
    tabela_exibicao["percentual_no_genero"]
    .round(2)
)

display(tabela_exibicao)


# =========================
# VALIDAÇÕES
# =========================

print("VALIDAÇÃO DO TOTAL GERAL")
print("Total da df_funcao:", len(df_funcao))

total_agrupamento = (
    distribuicao_genero_categoria["registros"].sum()
)

print("Total obtido no agrupamento:", total_agrupamento)

print(
    "Total geral confere:",
    total_agrupamento == len(df_funcao)
)

print("\nVALIDAÇÃO DOS TOTAIS POR GÊNERO")

validacao_genero = (
    distribuicao_genero_categoria
    .groupby("CL_GENERO")["registros"]
    .sum()
    .reset_index(name="total_calculado")
)

display(validacao_genero)

print("\nVALIDAÇÃO DOS PERCENTUAIS POR GÊNERO")

validacao_percentual = (
    distribuicao_genero_categoria
    .groupby("CL_GENERO")["percentual_no_genero"]
    .sum()
    .reset_index(name="soma_percentuais")
)

# Arredondamento apenas para exibição da validação
display(validacao_percentual.round(6))
````

### Interpretação dos resultados do CP4.5

**1. Distribuição das categorias**

A análise mostra que a categoria `ALIMENTOS` concentra o maior número de registros nos grupos de gênero analisados.

A distribuição relativa das categorias apresenta um padrão semelhante entre os grupos M e F.

**2. Comparação entre os grupos**

Embora existam diferenças nas contagens absolutas, a comparação percentual reduz o efeito das diferenças no volume total de registros de cada grupo.

Isso permite comparar a participação relativa das categorias, considerando o total de registros de cada grupo de gênero.

**3. Significado dos resultados**

Os valores obtidos representam ocorrências registradas na Base Varejo.

Essas contagens não devem ser interpretadas automaticamente como quantidade física de produtos adquiridos, valor gasto ou preferência dos consumidores.

**4. Qualidade dos dados**

Os registros completamente repetidos foram preservados porque não há evidência suficiente para classificá-los como erros.

Essa decisão pode influenciar as contagens observadas e deve ser considerada na interpretação dos resultados.

**5. Limitações da análise**

Uma mesma compra pode conter produtos de diferentes categorias.

Portanto, eventuais contagens de compras distintas por categoria não são necessariamente aditivas entre si.

Além disso, as distribuições observadas não permitem estabelecer relações causais entre gênero e comportamento de consumo.

### Unidade de análise

A unidade utilizada será o **registro da Base Varejo**, contado em cada combinação entre segmento e categoria de produto.

Cada linha representa uma ocorrência documentada, não necessariamente uma unidade física vendida.

### Procedimento

Serão investigados:

- A quantidade de registros em cada categoria por segmento.
- O total de registros pertencentes a cada segmento.
- A participação percentual de cada categoria dentro do respectivo segmento.
- A consistência dos totais obtidos em relação à base tratada.

### Cuidados metodológicos

````python
# CP4.6 — Distribuição das categorias por segmento

# Base: df_funcao (dados tratados na Sprint 3)

# 1. Contar registros por combinação de segmento e categoria

distribuicao_segmento_categoria = (
    df_funcao
    .groupby(["CL_SEG", "PR_CAT"])
    .size()
    .reset_index(name="QTD_REGISTROS")
)

print("ETAPA 1 — AGRUPAMENTO")
display(distribuicao_segmento_categoria)


# 2. Calcular o total de registros de cada segmento

totais_segmento = (
    df_funcao
    .groupby("CL_SEG")
    .size()
    .reset_index(name="TOTAL_SEGMENTO")
)

print("\nETAPA 2 — TOTAIS POR SEGMENTO")
display(totais_segmento)


# 3. Calcular percentuais dentro de cada segmento, sem arredondamento prévio

distribuicao_segmento_categoria = distribuicao_segmento_categoria.merge(
    totais_segmento,
    on="CL_SEG",
    how="left"
)

distribuicao_segmento_categoria["PERCENTUAL"] = (
    distribuicao_segmento_categoria["QTD_REGISTROS"]
    / distribuicao_segmento_categoria["TOTAL_SEGMENTO"]
    * 100
)


# 4. Organizar a tabela e arredondar somente para exibição

tabela_apresentacao = distribuicao_segmento_categoria.copy()

tabela_apresentacao["PERCENTUAL_EXIBICAO"] = (
    tabela_apresentacao["PERCENTUAL"].round(2)
)

tabela_apresentacao = tabela_apresentacao[
    [
        "CL_SEG",
        "PR_CAT",
        "QTD_REGISTROS",
        "TOTAL_SEGMENTO",
        "PERCENTUAL_EXIBICAO"
    ]
].sort_values(
    by=["CL_SEG", "PR_CAT"]
).reset_index(drop=True)

print("\nETAPA 4 — TABELA COMPARATIVA")
display(tabela_apresentacao)


# 5. Validar totais, segmentos, categorias e percentuais

print("\nETAPA 5 — VALIDAÇÕES")


# 5.1. Conferir o total agrupado de 830.000 registros

total_agrupado = distribuicao_segmento_categoria["QTD_REGISTROS"].sum()

print(
    "1. Total de registros dos agrupamentos:",
    total_agrupado,
    "| Esperado: 830000",
    "| OK:", total_agrupado == 830000
)


# 5.2. Conferir os segmentos A, B e C

segmentos_encontrados = set(
    distribuicao_segmento_categoria["CL_SEG"].unique()
)

segmentos_esperados = {"A", "B", "C"}

print(
    "2. Segmentos encontrados:",
    sorted(segmentos_encontrados),
    "| Esperados: A, B e C",
    "| OK:", segmentos_encontrados == segmentos_esperados
)


# 5.3. Conferir as sete categorias por segmento

categorias_por_segmento = (
    distribuicao_segmento_categoria
    .groupby("CL_SEG")["PR_CAT"]
    .nunique()
)

print("\n3. Quantidade de categorias por segmento:")
display(
    categorias_por_segmento
    .reset_index(name="QTD_CATEGORIAS")
)

validacao_sete_categorias = (
    categorias_por_segmento == 7
).all()

print(
    "Todos os segmentos possuem sete categorias:",
    validacao_sete_categorias
)


# 5.4. Conferir a soma de 100% dos percentuais por segmento

soma_percentuais = (
    distribuicao_segmento_categoria
    .groupby("CL_SEG")["PERCENTUAL"]
    .sum()
    .reset_index(name="SOMA_PERCENTUAL")
)

print("\n4. Soma percentual por segmento:")
display(soma_percentuais)

validacao_percentuais = (
    soma_percentuais["SOMA_PERCENTUAL"]
    .sub(100)
    .abs()
    .lt(1e-9)
    .all()
)

print(
    "Todos os segmentos somam aproximadamente 100%:",
    validacao_percentuais
)


# 5.5. Registrar o término das validações

print("\n5. Execução concluída sem erros.")
````

## CP4.7 — Investigação dos produtos e de suas frequências

### Pergunta de investigação

Quais produtos apresentam maior frequência nos registros da Base Varejo, e como essa distribuição muda quando consideramos compras e clientes distintos?

### Objetivo

Nas investigações anteriores, foram comparadas categorias de produtos por gênero e segmento.

Agora, o objetivo é aprofundar a análise no nível dos produtos individuais, utilizando seus códigos e nomes.

A proposta é distinguir diferentes medidas de frequência e verificar como a unidade de análise influencia a interpretação.

### Unidade de análise

Serão consideradas três medidas:

- **Registros:** quantidade de linhas associadas a cada produto.
- **Compras distintas:** quantidade de códigos `CO_ID` diferentes nos quais determinado produto aparece.
- **Clientes distintos:** quantidade de códigos `CL_ID` diferentes associados ao produto.

Essas medidas não são equivalentes.

### Procedimento

A investigação deverá identificar os produtos por `PR_ID` e `PR_NOME`, calcular as três medidas e comparar os resultados.

Também será necessário considerar o impacto potencial dos registros repetidos e dos produtos sem identificação completa.

### Cuidados metodológicos

O número de registros não representa automaticamente quantidade física vendida.

O número de compras distintas não representa faturamento ou volume de unidades.

Os registros repetidos serão preservados conforme as decisões justificadas na Sprint 2.

O produto identificado pelo código `107`, cuja descrição permanece não informada, deverá receber atenção específica na interpretação da qualidade dos dados.

````python
# CP4.7 — Investigação dos produtos e de suas frequências

# 1. Identificar os produtos e verificar a consistência entre código e nome

# Selecionar os campos necessários à investigação
produtos_base = df_funcao[
    ['PR_ID', 'PR_NOME', 'CO_ID', 'CL_ID']
].copy()

# Contar os nomes associados a cada código
consistencia_produtos = (
    produtos_base
    .groupby('PR_ID')['PR_NOME']
    .nunique(dropna=False)
    .reset_index(name='qtd_nomes_associados')
)

# Localizar códigos com mais de um nome
inconsistencias_produtos = consistencia_produtos[
    consistencia_produtos['qtd_nomes_associados'] > 1
]

print('ETAPA 1 — Identificação e consistência dos produtos')
print(
    f'Produtos distintos por PR_ID: '
    f'{produtos_base["PR_ID"].nunique(dropna=False)}'
)
print(
    f'Códigos associados a mais de um nome: '
    f'{len(inconsistencias_produtos)}'
)

if len(inconsistencias_produtos) > 0:
    display(inconsistencias_produtos)
else:
    print('Não foram encontrados códigos associados a mais de um nome.')

print()


# 2. Contar registros por produto

frequencia_registros = (
    produtos_base
    .groupby(['PR_ID', 'PR_NOME'], dropna=False)
    .size()
    .reset_index(name='registros')
    .sort_values('registros', ascending=False)
)

print('ETAPA 2 — Produtos com maior frequência por registros')
display(frequencia_registros.head(10))

print()


# 3. Contar compras distintas por produto

frequencia_compras = (
    produtos_base
    .groupby(['PR_ID', 'PR_NOME'], dropna=False)['CO_ID']
    .nunique()
    .reset_index(name='compras_distintas')
    .sort_values('compras_distintas', ascending=False)
)

print('ETAPA 3 — Produtos presentes em mais compras distintas')
display(frequencia_compras.head(10))

print()


# 4. Contar clientes distintos por produto

frequencia_clientes = (
    produtos_base
    .groupby(['PR_ID', 'PR_NOME'], dropna=False)['CL_ID']
    .nunique()
    .reset_index(name='clientes_distintos')
    .sort_values('clientes_distintos', ascending=False)
)

print('ETAPA 4 — Produtos associados a mais clientes distintos')
display(frequencia_clientes.head(10))

print()


# 5. Comparar as três medidas por produto

comparacao_produtos = (
    frequencia_registros
    .merge(
        frequencia_compras,
        on=['PR_ID', 'PR_NOME'],
        how='outer'
    )
    .merge(
        frequencia_clientes,
        on=['PR_ID', 'PR_NOME'],
        how='outer'
    )
)

# Calcular o ranking por registros
comparacao_produtos['rank_registros'] = (
    comparacao_produtos['registros']
    .rank(method='min', ascending=False)
    .astype('Int64')
)

# Calcular o ranking por compras distintas
comparacao_produtos['rank_compras'] = (
    comparacao_produtos['compras_distintas']
    .rank(method='min', ascending=False)
    .astype('Int64')
)

# Calcular o ranking por clientes distintos
comparacao_produtos['rank_clientes'] = (
    comparacao_produtos['clientes_distintos']
    .rank(method='min', ascending=False)
    .astype('Int64')
)

comparacao_produtos = comparacao_produtos.sort_values(
    ['rank_registros', 'rank_compras', 'rank_clientes']
)

print('ETAPA 5 — Comparação das três medidas')

display(
    comparacao_produtos[
        [
            'PR_ID',
            'PR_NOME',
            'registros',
            'compras_distintas',
            'clientes_distintos',
            'rank_registros',
            'rank_compras',
            'rank_clientes'
        ]
    ].head(15)
)

print()


# 6. Verificar qualidade dos dados e validar indicadores

# 6.1. Verificar repetições por produto, compra e cliente

duplicadas_produto_compra_cliente = produtos_base.duplicated(
    subset=['PR_ID', 'PR_NOME', 'CO_ID', 'CL_ID'],
    keep=False
)

qtd_linhas_repetidas = duplicadas_produto_compra_cliente.sum()

print('ETAPA 6 — Qualidade e validação')
print(
    'Linhas envolvidas em repetições de '
    'PR_ID + PR_NOME + CO_ID + CL_ID:',
    qtd_linhas_repetidas
)

print()


# 6.2. Comparar registros e compras distintas

registros_vs_compras = (
    produtos_base
    .groupby(['PR_ID', 'PR_NOME'], dropna=False)
    .agg(
        registros=('CO_ID', 'size'),
        compras_distintas=('CO_ID', 'nunique')
    )
    .reset_index()
)

registros_vs_compras['diferenca'] = (
    registros_vs_compras['registros']
    - registros_vs_compras['compras_distintas']
)

registros_vs_compras['registros_por_compra'] = (
    registros_vs_compras['registros']
    / registros_vs_compras['compras_distintas']
)

print('Produtos com maior diferença entre registros e compras distintas')

display(
    registros_vs_compras
    .sort_values('diferenca', ascending=False)
    .head(10)
)

print()


# 6.3. Conferir o produto de código 107

produto_107 = produtos_base[
    produtos_base['PR_ID'].astype(str).str.strip() == '107'
]

print('Produto PR_ID = 107')
print(f'Quantidade de registros: {len(produto_107)}')

if len(produto_107) > 0:
    print(
        'Compras distintas:',
        produto_107['CO_ID'].nunique()
    )

    print(
        'Clientes distintos:',
        produto_107['CL_ID'].nunique()
    )

    print(
        'Descrições encontradas:',
        produto_107['PR_NOME'].drop_duplicates().tolist()
    )

    display(produto_107.head(20))
else:
    print('O produto 107 não foi encontrado na base tratada.')

print()


# 6.4. Verificar nomes ausentes, vazios ou marcados como NÃO INFORMADO

produtos_sem_nome = produtos_base[
    produtos_base['PR_NOME'].isna()
    | produtos_base['PR_NOME'].astype(str).str.strip().isin(
        ['', 'nan', 'None', 'NÃO INFORMADO']
    )
]

print(
    'Registros com PR_NOME ausente, vazio ou NÃO INFORMADO:',
    len(produtos_sem_nome)
)

if len(produtos_sem_nome) > 0:
    display(
        produtos_sem_nome[
            ['PR_ID', 'PR_NOME']
        ]
        .drop_duplicates()
        .sort_values('PR_ID')
    )

print()


# 6.5. Resumir os líderes de cada indicador

top_registros = frequencia_registros.iloc[0]
top_compras = frequencia_compras.iloc[0]
top_clientes = frequencia_clientes.iloc[0]

print('RESUMO DA INVESTIGAÇÃO')

print(
    f'Maior número de registros: '
    f'{top_registros["PR_ID"]} - {top_registros["PR_NOME"]} '
    f'({top_registros["registros"]} registros)'
)

print(
    f'Maior número de compras distintas: '
    f'{top_compras["PR_ID"]} - {top_compras["PR_NOME"]} '
    f'({top_compras["compras_distintas"]} compras)'
)

print(
    f'Maior número de clientes distintos: '
    f'{top_clientes["PR_ID"]} - {top_clientes["PR_NOME"]} '
    f'({top_clientes["clientes_distintos"]} clientes)'
)
````

## CP4.8 — Análise temporal das compras

### Pergunta de investigação

Como as compras da Base Varejo estão distribuídas ao longo do período registrado na coluna `DATA`?

### Objetivo

Nas análises anteriores, investigamos categorias, gêneros, segmentos e produtos.

Neste checkpoint, vamos acrescentar a dimensão temporal para compreender como as operações de compra estão distribuídas entre diferentes meses.

### Unidade de análise

A principal unidade será a **compra distinta**, identificada por `CO_ID`.

Como uma mesma compra pode aparecer em várias linhas, a contagem de registros não será utilizada como equivalente ao número de compras.

### Procedimento

A investigação compreenderá:

- Validação do formato das datas;
- Identificação do período coberto pela base;
- Agrupamento das compras por mês;
- Contagem de compras distintas em cada período;
- Comparação dos resultados encontrados;
- Conferência da consistência das datas e dos indicadores.

### Cuidados metodológicos

A coluna `DATA` deverá ser convertida para um formato de data adequado, preservando os dados originais.

O período observado deverá ser considerado antes de realizar comparações entre meses.

Diferenças temporais serão interpretadas de maneira descritiva, sem afirmar sazonalidade ou relações causais que não estejam comprovadas pelos dados.

````python
# CP4.8 — Análise temporal das compras
# Unidade de análise: compra distinta (CO_ID)

import pandas as pd

# Trabalhar em uma cópia para preservar a base tratada
df_temporal = df_funcao.copy()

# 1. Preservar a coluna original
df_temporal["DATA_ORIGINAL"] = df_temporal["DATA"].copy()

# 2. Inspecionar o formato real antes da conversão
print("Amostra dos valores originais de DATA:")
print(df_temporal["DATA"].head(10))

print("\nTipo original da coluna DATA:")
print(df_temporal["DATA"].dtype)

print("\nValores ausentes em DATA:")
print(df_temporal["DATA"].isna().sum())

# 3. Converter DATA para datetime
# Formato dia/mês/ano identificado na base
df_temporal["DATA_DT"] = pd.to_datetime(
    df_temporal["DATA"],
    format="%d/%m/%Y",
    errors="coerce"
)

# 4. Conferir se houve falhas na conversão
falhas_conversao = df_temporal[
    df_temporal["DATA"].notna()
    & df_temporal["DATA_DT"].isna()
]

print("\nFalhas de conversão:")
print(len(falhas_conversao))

if len(falhas_conversao) > 0:
    print("\nExemplos de valores que não puderam ser convertidos:")
    print(
        falhas_conversao["DATA"]
        .drop_duplicates()
        .head(20)
    )

# 5. Identificar o período coberto pela base
data_inicial = df_temporal["DATA_DT"].min()
data_final = df_temporal["DATA_DT"].max()

print("\nPeríodo coberto pela base:")
print("Data inicial:", data_inicial)
print("Data final:", data_final)

# 6. Criar período mensal
df_temporal["MES"] = (
    df_temporal["DATA_DT"].dt.to_period("M")
)

# 7. Contar compras distintas por mês
compras_por_mes = (
    df_temporal
    .dropna(subset=["DATA_DT", "CO_ID"])
    .groupby("MES")["CO_ID"]
    .nunique()
    .reset_index(name="COMPRAS_DISTINTAS")
    .sort_values("MES")
)

# 8. Converter o período em texto para visualização
compras_por_mes["MES"] = (
    compras_por_mes["MES"].astype(str)
)

print("\nCompras distintas por mês:")
display(compras_por_mes)

# 9. Validação das compras
total_compras_periodo = (
    df_temporal.loc[
        df_temporal["DATA_DT"].notna(),
        "CO_ID"
    ].nunique()
)

total_soma_meses = (
    compras_por_mes["COMPRAS_DISTINTAS"].sum()
)

print("\nValidação das compras:")
print(
    "Compras distintas no período completo:",
    total_compras_periodo
)
print(
    "Soma das compras distintas mensais:",
    total_soma_meses
)

print(
    "Totais coincidem:",
    total_compras_periodo == total_soma_meses
)

# 10. Conferir a preservação da base tratada
print("\nValidação da base tratada:")
print(
    "Dimensões originais de df_funcao:",
    df_funcao.shape
)
print(
    "Dimensões da cópia temporal:",
    df_temporal.shape
)
print(
    "Total de registros preservado:",
    len(df_temporal) == len(df_funcao)
)
````

### Principais conclusões

#### 1. Qualidade e tratamento dos dados

A base original continha **830.000 registros e 14 colunas**.

Durante a limpeza, foram removidas quatro colunas completamente vazias, reduzindo a estrutura para **10 colunas**, sem excluir registros.

Também foram padronizadas **3.650 ocorrências de `#N/D` em `PR_CAT`** e outras **3.650 em `PR_NOME`**, correspondentes aos registros do produto de código 107.

As **96.553 ocorrências excedentes de linhas completamente repetidas**, identificadas no diagnóstico, foram preservadas porque não havia evidência suficiente para comprovar que representavam erros.

Essa decisão priorizou a preservação das informações e a rastreabilidade das transformações.

#### 2. Distribuição das categorias por gênero e segmento

A categoria `ALIMENTOS` apresentou a maior quantidade de registros da base, com **434.767 ocorrências**.

Na análise por gênero, sua participação foi semelhante entre os grupos: **52,38% em F** e **52,39% em M**.

O mesmo padrão foi observado entre os segmentos: **52,10% em A**, **52,40% em B** e **52,43% em C**.

Esses resultados mostram semelhança na composição percentual dos registros analisados, sem demonstrar, isoladamente, preferências individuais ou relações causais.

#### 3. Clientes, compras e segmentos

A base contém **1.000 clientes distintos e 18.471 compras distintas**.

Na análise por gênero, foram identificados **519 clientes e 9.615 compras no grupo F**, comparados a **481 clientes e 8.856 compras no grupo M**.

Na comparação por segmento, o código `B` apresentou **11.843 compras distintas**, seguido por `C`, com **5.136**, e `A`, com **1.492**.

Essas quantidades representam operações e clientes identificados na base, não valores gastos nem quantidades físicas de produtos.

Os códigos dos segmentos foram mantidos sem atribuição de características comerciais não documentadas.

#### 4. Diversidade de produtos por compra

A análise das **18.471 compras** identificou uma média de **39,71 códigos de produtos distintos por compra**, com mínimo de **1** e máximo de **81**.

Essa medida representa a diversidade de tipos de produtos associados às compras, e não a quantidade física de unidades vendidas.

Trata-se de uma variável quantitativa derivada dos identificadores disponíveis, cuja adequação ao critério institucional de estatísticas básicas permanece sujeita à avaliação.

#### 5. Frequência dos produtos e diferenças entre indicadores

A investigação dos produtos identificou **229 códigos `PR_ID` distintos**.

O produto com mais registros foi `PR_ID 93 — AGUA SANITARIA`, com **3.773 ocorrências**.

Por compras distintas, o líder foi `PR_ID 51 — PAPINHA INFANTIL`, presente em **3.322 compras**.

Por clientes distintos, destacou-se `PR_ID 155 — PATE`, associado a **975 clientes**.

Esses resultados demonstram que os rankings mudam conforme a unidade de análise utilizada. Assim, frequência de registros, presença em compras e abrangência entre clientes são medidas diferentes.

#### 6. Evolução temporal das compras

A análise temporal identificou **18.471 compras distintas**, registradas entre **4 de janeiro de 2019 e 8 de dezembro de 2022**.

Os totais anuais foram:

- **2019:** 4.483 compras;
- **2020:** 4.794 compras;
- **2021:** 5.505 compras;
- **2022:** 3.689 compras.

Foi observado crescimento entre 2019 e 2021, seguido de redução no total registrado em 2022.

A partir de setembro de 2022, ocorreu uma alteração acentuada na distribuição mensal, com **35 compras em setembro, 59 em outubro e 155 em novembro**.

Essa mudança precisa ser investigada antes de ser atribuída a fatores comerciais ou comportamentais, especialmente diante de possíveis diferenças na cobertura temporal dos dados.

Essas verificações poderiam melhorar a confiabilidade das análises sem exigir a exclusão de dados cuja inconsistência ainda não foi comprovada.

### Limitações da análise

A base não possui campos de **quantidade física vendida, preço, valor da venda, custo ou faturamento**.

Por esse motivo, não é possível calcular diretamente unidades comercializadas, receita, ticket médio, lucro ou margem financeira.

Além disso, os registros repetidos não puderam ser classificados definitivamente como erros, e os códigos de segmento não receberam interpretações comerciais sem documentação.

As análises apresentadas são descritivas e não demonstram relações causais entre gênero, segmento, produto e comportamento de compra.

Na dimensão temporal, a cobertura do período também deve ser considerada antes de estabelecer tendências de longo prazo.

Existe ainda uma ressalva metodológica no CP4.3: a medida de diversidade foi calculada a partir de identificadores de produtos e compras. Embora seja quantitativa e interpretável, não corresponde a uma coluna numérica original da base, e sua aderência à exigência específica da avaliação não está comprovada.
